```python
fizzbuzz_list = [
    #3と5の両方で割り切れる場合（15で割り切れる場合）
    "FizzBuzz" if i % 15 == 0 else
    #3で割り切れる場合
    "Fizz" if i % 3 == 0 else
    #5で割り切れる場合
    "Buzz" if i % 5 == 0 else
    #それ以外の場合は数値を文字列に変換
    str(i) 
    #1～100までの整数をループ
    for i in range(1,101)
]
#生成されたリストを全件出力
print(fizzbuzz_list)
```
