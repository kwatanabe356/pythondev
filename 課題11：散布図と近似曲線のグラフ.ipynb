```python
import numpy as np
import matplotlib.pyplot as plt

# 1. データの用意
x = np.array([62, 95, 83, 100, 72, 71, 55, 85, 67, 75])   # english
y = np.array([62, 80, 85, 95, 58, 76, 52, 84, 67, 71])    # math

# 2. 最小二乗法による近似直線の計算
# np.polyfit(x, y, 1) は 1次関数 (y = ax + b) の傾き a と切片 b を返します
a, b = np.polyfit(x, y, 1)
y_pred = a * x + b

# 3. グラフの描画
plt.figure(figsize=(8, 6))

# 散布図 (青色、ラベル: score)
plt.scatter(x, y, color='blue', label='score')

# 近似直線 (赤色、ラベル: lsm)
plt.plot(x, y_pred, color='red', label='lsm')

# 軸ラベルの設定
plt.xlabel('english')
plt.ylabel('math')

# 凡例の設定 (左上に表示)
plt.legend(loc='upper left')

# グラフの表示
plt.show()
```
