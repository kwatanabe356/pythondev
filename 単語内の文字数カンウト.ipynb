```python
#英単語を格納するための空のリストを作成
words = []

#a〜z をキー、0 を値とする辞書を作成
char_count = {chr(c): 0 for c in range(ord('a'), ord('z') + 1)}

#英単語の入力を受け付ける
while True:
    word = input("英単語を入力してください（未入力で終了）: ")

    #未入力で Enter が押されたら終了
    if word == "":
        break

    #入力された単語をリストに格納
    words.append(word)

    #単語を1文字ずつ確認してカウント
    for char in word:
        #小文字のアルファベットのみを対象にする
        if char in char_count:
            char_count[char] += 1

# ★ 出力前にABC順で並べ替え
words.sort()

#入力された英単語をABC順で表示
print("\n【入力された英単語一覧（ABC順）】")
for w in words:
    print(w)

#1文字以上出現した文字を表示
print("\n【文字の出現回数】")
for char, count in char_count.items():
    if count > 0:
        print(f"{char}が{count}個ありました")
```
